**Factor of Safety Prediction in Heterogeneous Slopes using Machine Learning**


#Assessing the stability of heterogeneous slopes is challenging because different geological materials have varying strength properties, and groundwater conditions further influence slope behaviour. Conventional numerical methods can accurately calculate the Factor of Safety (FoS), but evaluating many possible combinations of these conditions requires repeated simulations, making the process time-consuming and computationally demanding.

#The intended Machine Learning outcome is to develop a Regression model that learns the relationship between the geotechnical and groundwater parameters of a heterogeneous slope and its Factor of Safety. The model will provide rapid FoS predictions for new slope conditions without requiring a new numerical simulation for every scenario.




**Data Required**
The data set should contain geotechnical characteristics which include:

- Compressive strength of rock materials
- Hoek Brown parameters(mb,s and a)
- Tensile strength
- Friction angle
- Cohesion
- Pore water Pressure
- Factor of Safety(obtained from numerical slope stability analysis)

The process begins with the definition of a heterogeneous slope containing different geological materials. Different combinations of geotechnical and groundwater conditions are evaluated through numerical slope stability simulations to obtain corresponding FoS values.

The resulting simulation cases form the machine learning dataset. After data inspection and preprocessing, the geotechnical parameters are used as input features while FoS is selected as the prediction target.

The dataset is then divided into training and testing subsets. Regression algorithms can subsequently be trained using the training data and evaluated on previously unseen testing data using performance measures such as R², RMSE and MAE.

The trained model is then used to rapidly estimate FoS for new combinations of slope conditions.
